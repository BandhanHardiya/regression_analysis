Simpler Linear Regression


First we have to import esential libraries

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import statsmodels.api as sm
import seaborn as sns
```

Load the data into datafram from CSV with the help of pandas 

```python
data = pd.read_csv("1.01. Simple linear regression.csv")
```

now print some data of the data frame 

```python
data.head()
```

So, basically this dataset contain SAT score (entrance exam score) and college GPA (grade in college). In normal cases student with higger SAT score also get good score in college GPA so we are going to find regression equation for GPA (dependent) on SAT (indpendent) and also plot the best fitting line in scatter plot 

```python
data.describe()
```

```python
y = data['GPA']    #Dependent variable 
x1 = data['SAT']   #Independent variable 
```

Explore our data with the help of scatter plot 

```python
plt.scatter(x1, y)
plt.xlabel('SAT', fontsize = 20)
plt.ylabel('GPA', fontsize = 20)
plt.show()
```

As we can understand there is some positive correlation between SAT and GPA

we are going to use OLS (ordinal least square method) from statsmodel library to get required information to create a regression equation. OLS requires a constant (independent variable) to generate summery.

```python
x = sm.add_constant(x1)
result = sm.OLS(y, x).fit()
result.summary()
```

Now this table contain lot's of information which is very useful for any regression model. I suggest to read this artical to get detaild information about this summary https://towardsdatascience.com/simple-explanation-of-statsmodel-linear-regression-model-summary-35961919868b/

```python
plt.scatter(x1, y)
yhat = 0.0017 * x1 + 0.2750 
fig = plt.plot(x1, yhat, c = 'red', label = 'regression line')
plt.xlabel('SAT', fontsize = 20)
plt.ylabel('GPA', fontsize = 20)
plt.title('Simple Regression')
plt.show()
```
